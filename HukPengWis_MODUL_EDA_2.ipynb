## LATIHAN 1

```python
import numpy as np

harga = np.array ([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

**JAWABAN:**

Jika menggunakan list Python biasa, kode tersebut akan menghasilkan pesan error (TypeError). Python bawaan tidak bisa mengalikan list dengan bilangan desimal (float). Di Python standar, perkalian pada list hanya berfungsi dengan bilangan bulat (integer) untuk menggandakan/mengulang isi list (misal: [1, 2] * 2 menjadi [1, 2, 1, 2]). Sebaliknya, NumPy array memiliki fitur vectorization, yang memungkinkan operasi matematika diterapkan secara langsung ke setiap angka di dalam array tersebut sekaligus.

## LATIHAN 4

```python
import pandas as pd

df = pd.read_csv('data_kantin.csv')

print(df.isnull().sum())
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Ini adalah bagian dari logika pembersihan data (Data Cleaning):  

- Kolom terjual (Numerik): Jika jumlah terjual kosong, sangat logis untuk berasumsi bahwa pada saat itu barang tersebut tidak laku sama sekali, jadi masuk akal jika kita menggantinya dengan angka 0.
- Kolom menu (Kategorikal): Jika nama menu kosong, data tersebut kehilangan identitasnya. Kita tidak bisa menebak menu apa itu, dan baris data tanpa nama barang menjadi tidak berguna untuk dianalisis lebih lanjut (tidak bisa dikelompokkan). Oleh karena itu, baris tersebut lebih baik dibuang (drop).

## LATIHAN 5

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()

df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

**JAWABAN:**

Memastikan tipe data benar sangat krusial karena komputer memproses data sesuai tipenya. Jika kolom harga secara tidak sengaja terbaca sebagai teks (string / object) alih-alih angka (integer), kamu tidak akan bisa menjumlahkan, mengalikan, atau mencari rata-ratanya. Fungsi astype(int) memastikan data tersebut diakui sebagai angka valid. Mengenai jumlah baris setelah drop_duplicates(), baris hanya akan berkurang jika ada dua baris (atau lebih) yang datanya sama persis dari ujung ke ujung.

```python
import pandas as pd

df = pd.read_csv('data_kantin.csv')

laris = df[df['terjual'] > 20]
urut = df.sort_values(by='terjual', ascending=False) # filtering

# sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan

ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```


**JAWABAN:**

- Nasi Goreng: 12.000 * 23 = 276.000
- Es Teh: (4.000 * 40) + (4.000 * 35) = 300.000
- Mie Ayam: 10.000 * 0 (karena nilai kosong diganti 0) = 0
- Menu tertinggi adalah Es Teh (Rp 300.000).

Pengambilan Keputusan: Informasi ini membuktikan bahwa Es Teh adalah penyumbang pendapatan terbesar (produk paling laris).
